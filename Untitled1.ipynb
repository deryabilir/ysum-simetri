```python
import os
import requests
import zipfile
from google.colab import files

print("⏳ 100 İyi Huylu ve 100 Kötü Huylu lezyon görseli arşivden indiriliyor...")

# Klasörleri oluştur
os.makedirs("iyihuylu_nv_100", exist_ok=True)
os.makedirs("kotuhuylu_mel_100", exist_ok=True)

def gorselleri_indir(search_term, hedef_klasor, adet=100):
    url = f"https://api.isic-archive.com/api/v2/images/?limit={adet}&search=diagnosis%3A%22{search_term}%22"
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(url, headers=headers).json()

    sayac = 0
    for item in response.get('results', []):
        img_url = item['files']['thumbnail_256']['url']
        try:
            img_res = requests.get(img_url, headers=headers, timeout=5)
            if img_res.status_code == 200:
                sayac += 1
                dosya_yolu = os.path.join(hedef_klasor, f"{search_term}_{sayac}.jpg")
                with open(dosya_yolu, 'wb') as f:
                    f.write(img_res.content)
        except:
            continue
    print(f"✅ {hedef_klasor} klasörüne {sayac} adet görsel indirildi.")

# 100'er adet indir (Eğer zaten indirilmişlerse API isteklerini tekrar yapıp beklememek için kontrol ekleyebiliriz)
if len(os.listdir("iyihuylu_nv_100")) < 100:
    gorselleri_indir("melanocytic nevus", "iyihuylu_nv_100", 100)
else:
    print("✅ iyihuylu_nv_100 klasörü zaten dolu.")

if len(os.listdir("kotuhuylu_mel_100")) < 100:
    gorselleri_indir("melanoma", "kotuhuylu_mel_100", 100)
else:
    print("✅ kotuhuylu_mel_100 klasörü zaten dolu.")

# Klasörleri temiz bir şekilde zipfile ile sıkıştır
zip_dosya_adi = "lezyon_gorselleri_200.zip"
print("\n📦 Görseller ZIP arşivine ekleniyor...")
with zipfile.ZipFile(zip_dosya_adi, "w") as zipf:
    for root, dirs, files_list in os.walk("iyihuylu_nv_100"):
        for file in files_list:
            zipf.write(os.path.join(root, file), os.path.join("iyihuylu_nv_100", file))
    for root, dirs, files_list in os.walk("kotuhuylu_mel_100"):
        for file in files_list:
            zipf.write(os.path.join(root, file), os.path.join("kotuhuylu_mel_100", file))

print("\n🚀 İndirme tamamlandı! ZIP dosyası bilgisayarınıza iniyor...")
files.download(zip_dosya_adi)
```
